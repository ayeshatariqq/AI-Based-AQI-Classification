```python
import os, glob
import pandas as pd

pune_dir = 'data/pune'
print(os.listdir(pune_dir))

# Try to auto-pick the city-averaged file; if this finds nothing, set pune_path by hand
candidates = glob.glob(os.path.join(pune_dir, '*city*')) + glob.glob(os.path.join(pune_dir, '*City*'))
print("Candidates:", candidates)
```

```python
pune_path = candidates[0]   # or type the path manually
pune = pd.read_csv(pune_path)

print(pune.shape)
print(pune.columns.tolist())
print(pune.head())
print((pune.isnull().mean() * 100).round(1))
```

```python
import numpy as np

rename = {'pm25':'PM2.5','pm10':'PM10','no2':'NO2','so2':'SO2','co':'CO','o3':'O3'}
p = pune.rename(columns=rename)
p['date'] = pd.to_datetime(p['date'])

shared   = ['PM2.5','PM10','NO2','SO2','CO','O3']
sub_cols = ['sub_pm25','sub_pm10','sub_no2','sub_so2','sub_co','sub_o3']

complete = p.dropna(subset=shared + sub_cols + ['aqi']).copy()
print(len(p), '->', len(complete), 'complete rows')
print(complete['date'].min(), 'to', complete['date'].max())
```

```python
shared = ['PM2.5', 'PM10', 'NO2', 'SO2', 'CO', 'O3']

print("NaNs per shared column (train, before fill):")
print(X_train[shared].isnull().sum())

train_medians = X_train.median()
X_train = X_train.fillna(train_medians)
X_val   = X_val.fillna(train_medians)
X_test  = X_test.fillna(train_medians)

print("Remaining NaNs (train/val/test):",
      X_train.isnull().sum().sum(), X_val.isnull().sum().sum(), X_test.isnull().sum().sum())
```

```python
import numpy as np
import xgboost as xgb
from imblearn.over_sampling import SMOTE
from sklearn.metrics import accuracy_score, f1_score, classification_report, confusion_matrix

def evaluate_model(name, y_true, y_pred):
    print(f"\n{'='*50}\n{name}\n{'='*50}")
    print(f"Accuracy:   {accuracy_score(y_true, y_pred):.4f}")
    print(f"F1 (macro): {f1_score(y_true, y_pred, average='macro'):.4f}")
    print(classification_report(y_true, y_pred))

shared = ['PM2.5', 'PM10', 'NO2', 'SO2', 'CO', 'O3']

# --- Train the reduced-feature model (same tuned hyperparameters, SMOTE on train only) ---
y_train_enc = le.transform(y_train)
Xtr_sm, ytr_sm = SMOTE(random_state=42, k_neighbors=5).fit_resample(X_train[shared], y_train_enc)

xgb_reduced = xgb.XGBClassifier(
    n_estimators=200, max_depth=10, learning_rate=0.01,
    subsample=0.6, colsample_bytree=1.0, min_child_weight=1,
    random_state=42, eval_metric='mlogloss'
)
xgb_reduced.fit(Xtr_sm, ytr_sm)

# --- Reference: reduced model on the Indian val and test sets ---
for name, Xs, ys in [("Reduced model, India val",  X_val[shared],  y_val),
                     ("Reduced model, India test", X_test[shared], y_test)]:
    evaluate_model(name, ys, le.inverse_transform(xgb_reduced.predict(Xs)))

# --- Pune ---
y_true = complete['AQI_Bucket'].astype(str)
y_pred = le.inverse_transform(xgb_reduced.predict(complete[shared]))
evaluate_model("Reduced model, Pune (276 days)", y_true, y_pred)

order = ['Good', 'Satisfactory', 'Moderate', 'Poor', 'Very Poor', 'Severe']
idx = {c: i for i, c in enumerate(order)}
gap = np.abs(y_true.map(idx).values - pd.Series(y_pred).map(idx).values)
print("Within one category:", round((gap <= 1).mean(), 3))

main3 = ['Good', 'Satisfactory', 'Moderate']
print("Macro F1 (3 main classes):", round(f1_score(y_true, y_pred, labels=main3, average='macro'), 3))

pd.DataFrame(confusion_matrix(y_true, y_pred, labels=order), index=order, columns=order)
```

```python
from sklearn.metrics import f1_score, accuracy_score

order = ['Good','Satisfactory','Moderate','Poor','Very Poor','Severe']
idx = {c: i for i, c in enumerate(order)}
main3 = ['Good','Satisfactory','Moderate']
y_pune = complete['AQI_Bucket'].astype(str)

def fit_xgb(Xtr, ytr, use_smote):
    if use_smote:
        Xtr, ytr = SMOTE(random_state=42, k_neighbors=5).fit_resample(Xtr, ytr)
    m = xgb.XGBClassifier(n_estimators=200, max_depth=10, learning_rate=0.01,
                          subsample=0.6, colsample_bytree=1.0, min_child_weight=1,
                          random_state=42, eval_metric='mlogloss')
    m.fit(Xtr, ytr)
    return m

variants = {
    'A: 6 features + SMOTE (current)': (shared, True),
    'B: 6 features, no SMOTE':         (shared, False),
    'C: no CO/O3 + SMOTE':             (['PM2.5','PM10','NO2','SO2'], True),
    'D: no CO/O3, no SMOTE':           (['PM2.5','PM10','NO2','SO2'], False),
}

rows = []
for name, (feats, sm) in variants.items():
    m = fit_xgb(X_train[feats], y_train_enc, sm)
    p_in = le.inverse_transform(m.predict(X_test[feats]))
    p_pu = le.inverse_transform(m.predict(complete[feats]))

    t = y_pune.map(idx).values
    p = pd.Series(p_pu).map(idx).values
    wrong = t != p
    rows.append({
        'variant': name,
        'India_test_acc': round(accuracy_score(y_test, p_in), 3),
        'India_test_macroF1': round(f1_score(y_test, p_in, average='macro'), 3),
        'Pune_acc': round(accuracy_score(y_pune, p_pu), 3),
        'Pune_F1_main3': round(f1_score(y_pune, p_pu, labels=main3, average='macro', zero_division=0), 3),
        'Pune_over-predicted_share_of_errors': round(((p > t) & wrong).sum() / max(wrong.sum(), 1), 3),
    })

pd.DataFrame(rows)
```

```python
drop_cities = ['Ahmedabad','Chennai','Gurugram','Lucknow','Mumbai','Patna']

# 1. Daily max of rolling 8-hour averages for CO and O3, per city
hour = pd.read_csv('data/city_hour.csv', parse_dates=['Datetime'])
hour = hour[~hour['City'].isin(drop_cities)]
frames = []
for city, g in hour.groupby('City'):
    g = g.set_index('Datetime').sort_index()
    out = pd.DataFrame({
        'CO8': g['CO'].rolling('8h', min_periods=6).mean().resample('D').max(),
        'O38': g['O3'].rolling('8h', min_periods=6).mean().resample('D').max(),
    })
    out['City'] = city
    frames.append(out.reset_index().rename(columns={'Datetime': 'Date'}))
eight = pd.concat(frames)

# 2. Merge onto the daily file, swap in the 8-hour maxima, same cleaning as before
day = pd.read_csv('data/city_day.csv', parse_dates=['Date'])
day = day[~day['City'].isin(drop_cities)]
m = day.merge(eight, on=['City', 'Date'], how='left')
m['CO'], m['O3'] = m['CO8'], m['O38']

shared = ['PM2.5', 'PM10', 'NO2', 'SO2', 'CO', 'O3']
m = m.sort_values(['City', 'Date'])
m[shared] = m.groupby('City')[shared].transform(lambda x: x.ffill().bfill())
m = m.dropna(subset=shared + ['AQI_Bucket'])

# 3. Same per-city 70/15/15 chronological split
def split_city(g, test_frac=0.15, val_frac=0.15):
    g = g.sort_values('Date'); n = len(g)
    a = int(n * (1 - test_frac - val_frac)); b = int(n * (1 - test_frac))
    return g.iloc[:a], g.iloc[a:b], g.iloc[b:]

parts = [split_city(g) for _, g in m.groupby('City')]
tr = pd.concat([p[0] for p in parts]); va = pd.concat([p[1] for p in parts]); te = pd.concat([p[2] for p in parts])
print(tr.shape, va.shape, te.shape)

# 4. Retrain (same hyperparameters, SMOTE on train only) and score both datasets
model_h = fit_xgb(tr[shared], le.transform(tr['AQI_Bucket']), True)
p_in = le.inverse_transform(model_h.predict(te[shared]))
p_pu = le.inverse_transform(model_h.predict(complete[shared]))

t = y_pune.map(idx).values
p = pd.Series(p_pu).map(idx).values
wrong = t != p
pd.DataFrame([
    {'variant': 'A: daily-mean CO/O3 (before)', 'India_test_acc': 0.809, 'India_test_macroF1': 0.772,
     'Pune_acc': 0.725, 'Pune_F1_main3': 0.772, 'Pune_over_share': 0.908},
    {'variant': 'E: 8-hour-max CO/O3 (harmonized)',
     'India_test_acc': round(accuracy_score(te['AQI_Bucket'], p_in), 3),
     'India_test_macroF1': round(f1_score(te['AQI_Bucket'], p_in, average='macro'), 3),
     'Pune_acc': round(accuracy_score(y_pune, p_pu), 3),
     'Pune_F1_main3': round(f1_score(y_pune, p_pu, labels=main3, average='macro', zero_division=0), 3),
     'Pune_over_share': round(((p > t) & wrong).sum() / max(wrong.sum(), 1), 3)},
])
```

```python
import numpy as np, pandas as pd, joblib
import xgboost as xgb
from imblearn.over_sampling import SMOTE
from sklearn.metrics import accuracy_score, f1_score

le = joblib.load('models/label_encoder.pkl')

shared = ['PM2.5', 'PM10', 'NO2', 'SO2', 'CO', 'O3']
order  = ['Good', 'Satisfactory', 'Moderate', 'Poor', 'Very Poor', 'Severe']
idx    = {c: i for i, c in enumerate(order)}
main3  = ['Good', 'Satisfactory', 'Moderate']
drop_cities = ['Ahmedabad', 'Chennai', 'Gurugram', 'Lucknow', 'Mumbai', 'Patna']

def fit_xgb(Xtr, ytr, use_smote=True):
    if use_smote:
        Xtr, ytr = SMOTE(random_state=42, k_neighbors=5).fit_resample(Xtr, ytr)
    m = xgb.XGBClassifier(n_estimators=200, max_depth=10, learning_rate=0.01,
                          subsample=0.6, colsample_bytree=1.0, min_child_weight=1,
                          random_state=42, eval_metric='mlogloss')
    m.fit(Xtr, ytr)
    return m

# ---- Pune: only days where all six pollutants and sub-indices are present ----
pune = pd.read_csv('data/pune/city_averaged.csv')
pune = pune.rename(columns={'pm25': 'PM2.5', 'pm10': 'PM10', 'no2': 'NO2',
                            'so2': 'SO2', 'co': 'CO', 'o3': 'O3'})
sub_cols = ['sub_pm25', 'sub_pm10', 'sub_no2', 'sub_so2', 'sub_co', 'sub_o3']
complete = pune.dropna(subset=shared + sub_cols + ['aqi']).copy()
complete['AQI_Bucket'] = pd.cut(np.floor(complete['aqi'] + 0.5),
                                bins=[-np.inf, 50, 100, 200, 300, 400, np.inf], labels=order)
y_pune = complete['AQI_Bucket'].astype(str)
print("Pune complete days:", len(complete), "(expect 276)")

# ---- India: rebuild CO and O3 as the daily max of rolling 8-hour averages ----
hour = pd.read_csv('data/city_hour.csv', parse_dates=['Datetime'])
hour = hour[~hour['City'].isin(drop_cities)]
frames = []
for city, g in hour.groupby('City'):
    g = g.set_index('Datetime').sort_index()
    out = pd.DataFrame({
        'CO8': g['CO'].rolling('8h', min_periods=6).mean().resample('D').max(),
        'O38': g['O3'].rolling('8h', min_periods=6).mean().resample('D').max(),
    })
    out['City'] = city
    frames.append(out.reset_index().rename(columns={'Datetime': 'Date'}))
eight = pd.concat(frames)

day = pd.read_csv('data/city_day.csv', parse_dates=['Date'])
day = day[~day['City'].isin(drop_cities)]
m = day.merge(eight, on=['City', 'Date'], how='left')
m['CO'], m['O3'] = m['CO8'], m['O38']

# Same cleaning order as before: fill within city, then drop rows with no label
m = m.sort_values(['City', 'Date'])
m[shared] = m.groupby('City')[shared].transform(lambda x: x.ffill().bfill())
m = m.dropna(subset=['AQI', 'AQI_Bucket'])

def split_city(g, test_frac=0.15, val_frac=0.15):
    g = g.sort_values('Date'); n = len(g)
    a = int(n * (1 - test_frac - val_frac)); b = int(n * (1 - test_frac))
    return g.iloc[:a], g.iloc[a:b], g.iloc[b:]

parts = [split_city(g) for _, g in m.groupby('City')]
tr = pd.concat([p[0] for p in parts])
va = pd.concat([p[1] for p in parts])
te = pd.concat([p[2] for p in parts])
print(tr.shape[0], va.shape[0], te.shape[0], "rows (expect 11227, 2410, 2415)")

# Any gaps left are filled with TRAIN medians, as in the original pipeline
med = tr[shared].median()
for d in (tr, va, te):
    d[shared] = d[shared].fillna(med)

# ---- Retrain and score ----
model_h = fit_xgb(tr[shared], le.transform(tr['AQI_Bucket']), True)
p_in = le.inverse_transform(model_h.predict(te[shared]))
p_pu = le.inverse_transform(model_h.predict(complete[shared]))

t = y_pune.map(idx).values
p = pd.Series(p_pu).map(idx).values
wrong = t != p
pd.DataFrame([
    {'variant': 'A: daily-mean CO/O3 (before)', 'India_test_acc': 0.809, 'India_test_macroF1': 0.772,
     'Pune_acc': 0.725, 'Pune_F1_main3': 0.772, 'Pune_over_share': 0.908},
    {'variant': 'E: 8-hour-max CO/O3 (harmonized)',
     'India_test_acc': round(accuracy_score(te['AQI_Bucket'], p_in), 3),
     'India_test_macroF1': round(f1_score(te['AQI_Bucket'], p_in, average='macro'), 3),
     'Pune_acc': round(accuracy_score(y_pune, p_pu), 3),
     'Pune_F1_main3': round(f1_score(y_pune, p_pu, labels=main3, average='macro', zero_division=0), 3),
     'Pune_over_share': round(((p > t) & wrong).sum() / max(wrong.sum(), 1), 3)},
])
```
